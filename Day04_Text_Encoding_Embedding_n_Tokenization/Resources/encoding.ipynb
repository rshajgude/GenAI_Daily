## OneHotEncoder

https://scikit-learn.org/stable/modules/generated/sklearn.preprocessing.OneHotEncoder.html

```python

from sklearn.preprocessing import OneHotEncoder
import numpy
```

```python
documents = ["people watch movie", 
             "people watch cricket", 
             "people like movie", 
             "people like cricket"]
```

```python
tokens=[sentence.lower().split() for sentence in documents]
tokens
```

```python
all_words_sentence = [[sentence_words] for sentence_words in tokens ]
all_words_sentence
```

### understanding the below code as its list comprehenssion and nested for loop

```python
# understanding the below code as its list comprehenssion and nested for loop

all_words = [[word] for sentence_words in tokens for word in sentence_words]
all_words
```

```python
encoder = OneHotEncoder(sparse_output=False)
```

```python

encoder.fit(all_words)
```

```python
encoder.categories_
```

```python
encoder.categories_[0]
```

```python
print("Vocabulary:", encoder.categories_[0])
```

```python
tokens
```

### Word level encoding

```python
for sentence in tokens:
    print("Sentence:", sentence)
    for word in sentence:
        print("Word:", word)
        encoded_word = encoder.transform([[word]])
        print("Encoded Word:", encoded_word)
```

### Document / Sentence level encoding

```python

for sentence in tokens:
    encoded_sentence = encoder.transform([[word] for word in sentence])
    print("\nSentence:", sentence)
    print(encoded_sentence)
     
```

## BOW

https://scikit-learn.org/0.15/modules/generated/sklearn.feature_extraction.text.CountVectorizer.html

```python

from sklearn.feature_extraction.text import CountVectorizer
```

```python

documents = ["people watch movie and watch movie again", 
             "people watch cricket and watch cricket",
             "people like movie and like movie a lot",
             "people like cricket"]
     
```

```python
bow = CountVectorizer()
```

```python
bow_matrix=bow.fit_transform(documents)
```

```python
print("bow_matrix:", bow_matrix.toarray())
```

```python
print("Vocabulary:", bow.get_feature_names_out())
```

```python
sentence = ["people need to watch news"]
```

```python
bow_matrix=bow.transform(sentence)
print("bow_matrix:", bow_matrix.toarray())
```

##### Observe - "need" "to" "news" words are totally missing in BOW representation

```python
print("Vocabulary:", bow.get_feature_names_out())
```

### TD-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python

documents = ["people watch movie and watch movie again", 
             "people watch cricket and watch cricket",
             "people like movie and like movie a lot",
             "people like cricket"]
```

```python

tf_idf = TfidfVectorizer()
```

```python
tf_idf_vector = tf_idf.fit_transform(documents)
```

```python
tf_idf_vector
```

```python
# call on tf-idf object to get the feature names (vocabulary) of the documents
print("Vocabulary:", tf_idf.get_feature_names_out())
```

```python
# call to_array on transformed by tf-idf model to get the tf-idf matrix of the documents
print("Matrix\n", tf_idf_vector.toarray())
```
