```python
from dotenv import load_dotenv
from groq import Groq
import os

load_dotenv()
client = Groq(api_key=os.getenv("GROQ_API_KEY"))

models = client.models.list()

for model in models.data:
    print(model.id)
```

```python
# Model Selected qwen/qwen3.6-27b
llm = ChatGroq(
    model="qwen/qwen3.6-27b")

```

```python
## this message which we are passsing it is only called prompt or input 
messages = [
    (
        "system",
        "You are a helpful assistant that translates English to Hindi. Translate the user sentence.",
    ),
    ("human", "I love programming."),
]
```

```python
# giving prompt to the model and getting the response from the model
ai_msg = llm.invoke(messages)
```

```python
# Response from the model not interpreted by the model, it is just a response from the model
for line in ai_msg.content.split("\n"):
    print(line)
```

### Gemini Image to text

```python
import base64
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import HumanMessage

# 1. Image ko base64 me convert karo
def encode_image(image_path):
    with open(image_path, "rb") as image_file:
        return base64.b64encode(image_file.read()).decode("utf-8")

base64_image = encode_image("/Users/rajkumar/GenAI_Daily/Day18_Access_LLM_Using_diff_API/ganapati_bappa_morya.jpeg")
```

```python
# 2. Gemini model initialize karo
llm = ChatGoogleGenerativeAI(
    model="models/gemini-3.1-flash-image",
    google_api_key=GOOGLE_API_KEY
)
```

```python
# 3. Text + image message banao
message = HumanMessage(
    content=[
        {
            "type": "text",
            "text": "Describe this image in detail."
        },
        {
            "type": "image_url",
            "image_url": f"data:image/png;base64,{base64_image}"
        }
    ]
)
```

```python
# 4. Model invoke karo
response = llm.invoke([message])
```

```python
from langchain_ollama import ChatOllama
from langchain_core.messages import SystemMessage, HumanMessage
```

```python
# 1. Create local Ollama LLM
llm = ChatOllama(
    model="qwen2.5-coder:1.5b",
    temperature=0
)
```

```python
# 2. Create messages
messages = [
    SystemMessage(content="You are a helpful AI assistant."),
    HumanMessage(content="Explain RAG in simple words in AI")
]
```

```python
# 3. Invoke model
response = llm.invoke(messages)
```

```python
print(response.content)
```

```python
for sentence in response.content.split("."):
    print(sentence.strip())
```
