```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI
```

```python
load_dotenv()
llm = ChatGoogleGenerativeAI(model="models/gemma-4-26b-a4b-it")

def ask(question: str) -> str:
    response = llm.invoke(question)
    return response.content
```

```python
# Usage
answer = ask("List of countries in Africa.")
print(answer)
```

```python
for line in answer:
    print(line)
```
